### Credit Score Classification using Random Forest
In this notebook, we build a pipeline to classify individuals' credit scores into categories (**Poor**, **Standard**, **Good**) using key financial indicators like income, outstanding debt, loan counts, interest rates, and payment history behaviors.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_2_component = None # clear unused
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. Generate a realistic synthetic Credit Score dataset
np.random.seed(42)
n_samples = 1000

data = {
    'Age': np.random.randint(18, 70, size=n_samples),
    'Annual_Income': np.random.uniform(20000, 150000, size=n_samples),
    'Num_Bank_Accounts': np.random.randint(1, 10, size=n_samples),
    'Num_Credit_Card': np.random.randint(1, 12, size=n_samples),
    'Interest_Rate': np.random.uniform(1, 35, size=n_samples),
    'Num_of_Loan': np.random.randint(0, 10, size=n_samples),
    'Delay_from_due_date': np.random.randint(0, 40, size=n_samples),
    'Num_of_Delayed_Payment': np.random.randint(0, 25, size=n_samples),
    'Outstanding_Debt': np.random.uniform(100, 5000, size=n_samples),
    'Credit_Utilization_Ratio': np.random.uniform(20, 50, size=n_samples),
    'Payment_of_Min_Amount': np.random.choice(['Yes', 'No', 'NM'], size=n_samples)
}

df_credit = pd.DataFrame(data)

# Programmatically assign structured targets (Credit Score) based on risk factors
def assign_credit_score(row):
    score_points = 0
    if row['Outstanding_Debt'] > 2500: score_points += 3
    if row['Delay_from_due_date'] > 20: score_points += 2
    if row['Interest_Rate'] > 20: score_points += 2
    if row['Num_of_Delayed_Payment'] > 12: score_points += 2
    if row['Credit_Utilization_Ratio'] > 40: score_points += 1
    if row['Payment_of_Min_Amount'] == 'Yes': score_points -= 1

    if score_points >= 6:
        return 'Poor'
    elif score_points >= 3:
        return 'Standard'
    else:
        return 'Good'

df_credit['Credit_Score'] = df_credit.apply(assign_credit_score, axis=1)

# Display preview
display(df_credit.head())
print("\nCredit Score Distribution:")
print(df_credit['Credit_Score'].value_value_counts() if hasattr(df_credit['Credit_Score'], 'value_value_counts') else df_credit['Credit_Score'].value_counts())
```

### 2. Feature Engineering & Preprocessing
We will encode our categorical fields and scale the numeric features to prepare the datasets for the classifier.

```python
# Encode categorical features
le_min_pay = LabelEncoder()
df_credit['Payment_of_Min_Amount'] = le_min_pay.fit_transform(df_credit['Payment_of_Min_Amount'])

# Split features and target
X_c = df_credit.drop(columns=['Credit_Score'])
y_c = df_credit['Credit_Score']

# Encode target
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y_c) # Maps Good/Poor/Standard to numbers

# Split into training and testing sets
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_c, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

# Scale numerical variables
scaler_c = StandardScaler()
X_train_scaled_c = scaler_c.fit_transform(X_train_c)
X_test_scaled_c = scaler_c.transform(X_test_c)

print(f"Training set shape: {X_train_scaled_c.shape}")
print(f"Testing set shape: {X_test_scaled_c.shape}")
```

### 3. Model Training & Evaluation
We fit a Random Forest Classifier and evaluate its predictions using metric reports and confusion matrices.

```python
# Train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_scaled_c, y_train_c)

# Make predictions
y_pred_c = rf_model.predict(X_test_scaled_c)

# Evaluate
acc = accuracy_score(y_test_c, y_pred_c)
print(f"Model Accuracy: {acc * 100:.2f}%\n")

print("Classification Report:")
print(classification_report(y_test_c, y_pred_c, target_names=le_target.classes_))

# Plot Confusion Matrix
plt.figure(figsize=(8, 6))
cm_c = confusion_matrix(y_test_c, y_pred_c)
sns.heatmap(cm_c, annot=True, fmt='d', cmap='Blues', xticklabels=le_target.classes_, yticklabels=le_target.classes_)
plt.title('Credit Score Classification Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
