```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Generate synthetic 1D dataset from two different Gaussians
np.random.seed(42)
class1 = np.random.normal(loc=2.0, scale=1.0, size=150)
class2 = np.random.normal(loc=7.0, scale=1.5, size=150)
X = np.concatenate([class1, class2])

# 2. Implement the EM Algorithm from Scratch for 2 Components
def em_gaussian_mixture(X, max_iter=100, tol=1e-6):
    # Initialize parameters randomly
    mu1, mu2 = 1.0, 5.0
    sigma1, sigma2 = 1.0, 1.0
    pi1 = 0.5
    pi2 = 0.5

    n = len(X)
    log_likelihood_history = []

    for i in range(max_iter):
        # --- Expectation Step (E-step) ---
        # Calculate probability density under current parameters
        pdf1 = norm.pdf(X, loc=mu1, scale=sigma1)
        pdf2 = norm.pdf(X, loc=mu2, scale=sigma2)

        # Marginal probability / denominator
        total_density = pi1 * pdf1 + pi2 * pdf2

        # Compute responsibilities (membership weights)
        gamma1 = (pi1 * pdf1) / total_density
        gamma2 = (pi2 * pdf2) / total_density

        # --- Maximization Step (M-step) ---
        # Sum of responsibilities
        n1 = np.sum(gamma1)
        n2 = np.sum(gamma2)

        # Update means
        mu1 = np.sum(gamma1 * X) / n1
        mu2 = np.sum(gamma2 * X) / n2

        # Update variances (standard deviations)
        sigma1 = np.sqrt(np.sum(gamma1 * (X - mu1)**2) / n1)
        sigma2 = np.sqrt(np.sum(gamma2 * (X - mu2)**2) / n2)

        # Update mixing coefficients
        pi1 = n1 / n
        pi2 = n2 / n

        # Calculate log-likelihood to track convergence
        log_likelihood = np.sum(np.log(total_density))
        log_likelihood_history.append(log_likelihood)

        # Check for convergence
        if i > 0 and abs(log_likelihood_history[-1] - log_likelihood_history[-2]) < tol:
            print(f"Converged early at iteration {i}")
            break

    return (mu1, sigma1, pi1), (mu2, sigma2, pi2), log_likelihood_history

# 3. Fit GMM using EM
comp1, comp2, history = em_gaussian_mixture(X)

print("--- EM Algorithm Final Parameters ---")
print(f"Component 1: Mean = {comp1[0]:.4f}, Std Dev = {comp1[1]:.4f}, Weight = {comp1[2]:.4f}")
print(f"Component 2: Mean = {comp2[0]:.4f}, Std Dev = {comp2[1]:.4f}, Weight = {comp2[2]:.4f}\n")

# 4. Visualize Fitted Gaussians over the Original Data Histogram
plt.figure(figsize=(12, 6))
x_axis = np.linspace(min(X) - 2, max(X) + 2, 1000)

# Plot histogram
plt.hist(X, bins=30, density=True, color='lightgray', edgecolor='white', label='Data Histogram')

# Plot estimated distributions
pdf_est1 = comp1[2] * norm.pdf(x_axis, loc=comp1[0], scale=comp1[1])
pdf_est2 = comp2[2] * norm.pdf(x_axis, loc=comp2[0], scale=comp2[1])
plt.plot(x_axis, pdf_est1, color='royalblue', linewidth=2.5, label='Fitted Gaussian 1')
plt.plot(x_axis, pdf_est2, color='darkorange', linewidth=2.5, label='Fitted Gaussian 2')
plt.plot(x_axis, pdf_est1 + pdf_est2, color='forestgreen', linewidth=2, linestyle='--', label='Combined GMM Fit')

plt.title('Expectation-Maximization (EM) GMM Fit')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
