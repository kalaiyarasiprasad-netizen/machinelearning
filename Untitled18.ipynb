### 1. Import Libraries and Load Iris Dataset

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report

# Load the dataset
iris = load_iris()
# Using two features for 2D visualization (Sepal Length and Petal Length)
X = iris.data[:, [0, 2]]
y = iris.target

# Standard Perceptron is a binary classifier.
# Let's convert this to a binary classification problem: Setosa (0) vs Versicolor (1)
# We'll filter out Virginica (class 2) for this demonstration
binary_mask = y != 2
X = X[binary_mask]
y = y[binary_mask]

print(f"Dataset shape: {X.shape}")
print(f"Classes: {np.unique(y)} (0: Setosa, 1: Versicolor)")
```
