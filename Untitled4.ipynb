````python
import numpy as np
import pandas as pd

# --- Custom ID3 Decision Tree Implementation ---

class ID3Node:
    def __init__(self, feature=None, results=None, child_nodes=None):
        self.feature = feature           # Feature name to split on
        self.results = results           # If leaf node, contains the class label
        self.child_nodes = child_nodes or {}  # Dictionary to hold branch_value -> ID3Node mappings

def calculate_entropy(target_col):
    elements, counts = np.unique(target_col, return_counts=True)
    entropy = 0.0
    for i in range(len(elements)):
        p = counts[i] / np.sum(counts)
        entropy -= p * np.log2(p)
    return entropy

def calculate_info_gain(data, split_attribute_name, target_attribute_name):
    total_entropy = calculate_entropy(data[target_attribute_name])

    vals, counts = np.unique(data[split_attribute_name], return_counts=True)
    weighted_entropy = 0.0
    for i in range(len(vals)):
        subset = data[data[split_attribute_name] == vals[i]]
        subset_entropy = calculate_entropy(subset[target_attribute_name])
        weighted_entropy += (counts[i] / np.sum(counts)) * subset_entropy

    information_gain = total_entropy - weighted_entropy
    return information_gain

def build_id3_tree(data, original_data, features, target_attribute_name):
    # If all target values are the same, return a leaf node with this value
    unique_targets = np.unique(data[target_attribute_name])
    if len(unique_targets) <= 1:
        return ID3Node(results=unique_targets[0])

    # If dataset is empty, return the most common target in original dataset
    elif len(data) == 0:
        vals, counts = np.unique(original_data[target_attribute_name], return_counts=True)
        return ID3Node(results=vals[np.argmax(counts)])

    # If features list is empty, return the most common target in current dataset
    elif len(features) == 0:
        vals, counts = np.unique(data[target_attribute_name], return_counts=True)
        return ID3Node(results=vals[np.argmax(counts)])

    # Otherwise, grow the tree
    else:
        item_values = [calculate_info_gain(data, feature, target_attribute_name) for feature in features]
        best_feature_index = np.argmax(item_values)
        best_feature = features[best_feature_index]

        tree = ID3Node(feature=best_feature)
        remaining_features = [f for f in features if f != best_feature]

        for value in np.unique(original_data[best_feature]):
            sub_data = data[data[best_feature] == value]

            if len(sub_data) == 0:
                vals, counts = np.unique(original_data[target_attribute_name], return_counts=True)
                tree.child_nodes[value] = ID3Node(results=vals[np.argmax(counts)])
            else:
                tree.child_nodes[value] = build_id3_tree(sub_data, original_data, remaining_features, target_attribute_name)

        return tree

def print_tree(node, depth=0):
    indent = "  " * depth
    if node.results is not None:
        print(f"{indent}--> Predict: {node.results}")
    else:
        print(f"{indent}[Split on: {node.feature}]")
        for value, child in node.child_nodes.items():
            print(f"{indent}  Value: {value}")
            print_tree(child, depth + 2)

def classify(node, sample):
    if node.results is not None:
        return node.results

    attr_val = sample.get(node.feature)
    if attr_val in node.child_nodes:
        return classify(node.child_nodes[attr_val], sample)
    else:
        return "Unable to classify (unseen attribute path)"

# --- Demo Dataset (Play Tennis) ---
tennis_data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rainy', 'Rainy', 'Rainy', 'Overcast', 'Sunny', 'Sunny', 'Rainy', 'Sunny', 'Overcast', 'Overcast', 'Rainy'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'High'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak', 'Strong'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}

df_tennis = pd.DataFrame(tennis_data)
print("--- Play Tennis Training Dataset ---")
display(df_tennis)

# Train ID3 Tree
features_list = list(df_tennis.columns[:-1])
tree_root = build_id3_tree(df_tennis, df_tennis, features_list, 'PlayTennis')

print("\n--- Constructed Decision Tree Structure ---")
print_tree(tree_root)

# --- Classify a New Sample ---
new_sample = {
    'Outlook': 'Sunny',
    'Temperature': 'Cool',
    'Humidity': 'Normal',
    'Wind': 'Strong'
}

prediction = classify(tree_root, new_sample)
print(f"\n--- Classification Outcome ---")
print(f"New Sample: {new_sample}")
print(f"Predicted Class: {prediction}")
````

```markdown
# Artificial Neural Network with Backpropagation Algorithm

This notebook contains a complete, from-scratch implementation of a Multi-Layer Perceptron (MLP) neural network utilizing **Backpropagation** for weight updates.

### Mathematical Foundations
- **Activation Function (Sigmoid)**:
  $$\sigma(z) = \frac{1}{1 + e^{-z}}$$
- **Derivative of Sigmoid**:
  $$\sigma'(z) = \sigma(z)(1 - \sigma(z))$$
- **Loss Function (Mean Squared Error)**:
  $$E = \frac{1}{2}\sum (y - \hat{y})^2$$
```

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- Custom ANN Implementation ---
class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.1):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.lr = learning_rate

        # Initialize weights and biases randomly
        self.W1 = np.random.randn(self.input_size, self.hidden_size)
        self.b1 = np.zeros((1, self.hidden_size))
        self.W2 = np.random.randn(self.hidden_size, self.output_size)
        self.b2 = np.zeros((1, self.output_size))

    def sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-z))

    def sigmoid_derivative(self, a):
        # 'a' represents the activated output: sigmoid(z)
        return a * (1.0 - a)

    def forward(self, X):
        # Hidden layer forward path
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = self.sigmoid(self.z1)

        # Output layer forward path
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = self.sigmoid(self.z2)
        return self.a2

    def backward(self, X, y, output):
        # Error in output
        error_output = y - output
        # Output layer gradient
        delta_output = error_output * self.sigmoid_derivative(output)

        # Error propagated back to hidden layer
        error_hidden = np.dot(delta_output, self.W2.T)
        # Hidden layer gradient
        delta_hidden = error_hidden * self.sigmoid_derivative(self.a1)

        # Update weights and biases
        self.W2 += np.dot(self.a1.T, delta_output) * self.lr
        self.b2 += np.sum(delta_output, axis=0, keepdims=True) * self.lr

        self.W1 += np.dot(X.T, delta_hidden) * self.lr
        self.b1 += np.sum(delta_hidden, axis=0, keepdims=True) * self.lr

    def train(self, X, y, epochs=10000):
        loss_history = []
        for epoch in range(epochs):
            output = self.forward(X)
            self.backward(X, y, output)
            loss = np.mean(np.square(y - output))
            loss_history.append(loss)

            if (epoch + 1) % (epochs // 10) == 0:
                print(f"Epoch {epoch + 1}/{epochs} - Loss: {loss:.6f}")
        return loss_history
````

````python
# --- Demo on XOR Classification Task ---
# Features: X1, X2
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# True targets
y = np.array([
    [0],
    [1],
    [1],
    [0]
])

print("--- Training Data ---")
print(pd.DataFrame(np.hstack((X, y)), columns=['X1', 'X2', 'Y']))

# Initialize Neural Network (2 inputs, 4 hidden units, 1 output)
np.random.seed(42)
nn = NeuralNetwork(input_size=2, hidden_size=4, output_size=1, learning_rate=0.2)

# Train
print("\n--- Starting Training ---")
loss_history = nn.train(X, y, epochs=15000)

# Plot Loss Curve
plt.figure(figsize=(8, 4))
plt.plot(loss_history, color='blue')
plt.title("Training Loss Decay")
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error (MSE)")
plt.grid(True)
plt.show()
````

````python
# --- Testing and Verification ---
predictions = nn.forward(X)
print("\n--- Verification Results ---")
for i in range(len(X)):
    predicted_val = predictions[i][0]
    rounded_val = 1 if predicted_val >= 0.5 else 0
    print(f"Input: {X[i]} -> Target: {y[i][0]} | Prob: {predicted_val:.4f} -> Prediction: {rounded_val}")
````
