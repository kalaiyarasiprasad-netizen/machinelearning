```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

# 1. Generate non-linear synthetic data
np.random.seed(42)
X = 6 * np.random.rand(100, 1) - 3
y = 0.5 * X**3 + X**2 + 2 + np.random.randn(100, 1) # Cubic relationship with noise

# Sort data for smooth visualization plotting
sort_idx = np.argsort(X.flatten())
X_sorted = X[sort_idx]
y_sorted = y[sort_idx]

# 2. Linear Regression
linear_reg = LinearRegression()
linear_reg.fit(X, y)
y_pred_linear = linear_reg.predict(X_sorted)

# 3. Polynomial Regression (Degree = 3)
poly_features = PolynomialFeatures(degree=3, include_bias=False)
X_poly = poly_features.fit_transform(X)
X_poly_sorted = poly_features.transform(X_sorted)

poly_reg = LinearRegression()
poly_reg.fit(X_poly, y)
y_pred_poly = poly_reg.predict(X_poly_sorted)

# 4. Compute Metrics
mse_linear = mean_squared_error(y_sorted, y_pred_linear)
r2_linear = r2_score(y_sorted, y_pred_linear)

mse_poly = mean_squared_error(y_sorted, y_pred_poly)
r2_poly = r2_score(y_sorted, y_pred_poly)

print("--- Metric Comparison ---")
print(f"Linear Regression     -> MSE: {mse_linear:.4f} | R^2: {r2_linear:.4f}")
print(f"Polynomial (Degree 3) -> MSE: {mse_poly:.4f}  | R^2: {r2_poly:.4f}\n")

# 5. Visualization
plt.figure(figsize=(12, 7))
plt.scatter(X, y, color='gray', alpha=0.6, label='Data points')
plt.plot(X_sorted, y_pred_linear, color='crimson', linestyle='--', linewidth=2, label=f'Linear Fit (R² = {r2_linear:.2f})')
plt.plot(X_sorted, y_pred_poly, color='forestgreen', linewidth=3, label=f'Polynomial Fit d=3 (R² = {r2_poly:.2f})')
plt.title('Comparison: Linear vs. Polynomial Regression')
plt.xlabel('Feature X')
plt.ylabel('Target y')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
