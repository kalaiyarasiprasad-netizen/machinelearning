```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Generate synthetic continuous linear data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)        # 100 random feature values between 0 and 2
y = 4 + 3 * X + np.random.randn(100, 1) # Target variable with some Gaussian noise

# 2. Initialize and fit the Linear Regression Model
lin_reg = LinearRegression()
lin_reg.fit(X, y)

# 3. Make predictions across the training domain
y_pred = lin_reg.predict(X)

# 4. Extract and print coefficients and performance metrics
slope = lin_reg.coef_[0][0]
intercept = lin_reg.intercept_[0]
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("--- Linear Regression Performance ---")
print(f"Estimated Slope (w1): {slope:.4f}")
print(f"Estimated Intercept (w0): {intercept:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Coefficient of Determination (R^2): {r2:.4f}\n")

# 5. Visualize the data points and the regression line
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='royalblue', alpha=0.7, label='Data Points (y = 4 + 3X + Noise)')
plt.plot(X, y_pred, color='crimson', linewidth=2.5, label=f'Regression Line (y = {intercept:.2f} + {slope:.2f}X)')
plt.title('Simple Linear Regression Fit')
plt.xlabel('Feature X')
plt.ylabel('Target y')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
