```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Synthesize a realistic Mobile Specifications Dataset
np.random.seed(42)
n_samples = 1000

ram = np.random.choice([2, 3, 4, 6, 8, 12, 16], size=n_samples, p=[0.1, 0.1, 0.2, 0.2, 0.2, 0.15, 0.05])
storage = ram * np.random.choice([8, 16, 32], size=n_samples, p=[0.4, 0.4, 0.2]) # Storage depends loosely on RAM
battery = np.random.randint(3000, 6000, size=n_samples)
screen_size = np.round(np.random.uniform(5.5, 6.9, size=n_samples), 1)
primary_camera = np.random.choice([12, 48, 64, 108], size=n_samples, p=[0.3, 0.4, 0.2, 0.1])

# Generate price using a non-linear relationship with added noise
base_price = 100
price = (
    base_price +
    (ram * 35) +
    (storage * 2.2) +
    ((battery - 3000) * 0.05) +
    ((screen_size - 5.5) * 50) +
    (primary_camera * 1.5) +
    np.random.normal(0, 25, size=n_samples)
)

# Keep prices positive
price = np.clip(price, 50, None)

# Create DataFrame
df_mobiles = pd.DataFrame({
    'RAM_GB': ram,
    'Storage_GB': storage,
    'Battery_mAh': battery,
    'Screen_Size_Inches': screen_size,
    'Primary_Camera_MP': primary_camera,
    'Price_USD': price
})

display(df_mobiles.head())
```
