```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split

# --- Custom K-NN Classification ---
class KNNClassifier:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = np.array(X)
        self.y_train = np.array(y)

    def _euclidean_distance(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))

    def predict(self, X_test):
        X_test = np.array(X_test)
        return np.array([self._predict_single(x) for x in X_test])

    def _predict_single(self, x_test_sample):
        # 1. Compute distances to all training samples
        distances = [self._euclidean_distance(x_test_sample, x_train_sample) for x_train_sample in self.X_train]

        # 2. Get indices of the k nearest neighbors
        k_neighbor_indices = np.argsort(distances)[:self.k]

        # 3. Get the labels of these neighbors
        k_neighbor_labels = [self.y_train[i] for i in k_neighbor_indices]

        # 4. Perform majority voting
        most_common = Counter(k_neighbor_labels).most_common(1)
        return most_common[0][0]

# --- Create a Synthetic Dataset ---
X_data, y_data = make_blobs(n_samples=150, n_features=2, centers=3, cluster_std=1.2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2, random_state=42)

# --- Initialize and Train K-NN Model ---
knn = KNNClassifier(k=5)
knn.fit(X_train, y_train)

# --- Make Predictions and Evaluate ---
y_pred = knn.predict(X_test)
accuracy = np.mean(y_pred == y_test) * 100
print(f"--- K-NN Classification Results (k=5) ---")
print(f"Overall Model Test Accuracy: {accuracy:.2f}%")

# --- Classify a Custom New Sample ---
custom_sample = np.array([[-2.5, 4.5]])
custom_pred = knn.predict(custom_sample)
print(f"\nPrediction for custom sample {custom_sample[0]}: Class {custom_pred[0]}")

# --- Plotting the Boundaries and Classification ---
plt.figure(figsize=(10, 6))
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='viridis', marker='o', alpha=0.5, label='Training Data')
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_pred, cmap='viridis', marker='x', s=100, linewidths=2, label='Predicted Test Data')
plt.scatter(custom_sample[:, 0], custom_sample[:, 1], color='red', marker='*', s=200, label='Custom New Sample')
plt.title("K-NN Classification visualization (o = Train, x = Test Prediction, * = New Sample)")
plt.xlabel("Feature Space X1")
plt.ylabel("Feature Space X2")
plt.grid(True)
plt.legend()
plt.show()
```
