```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Generate realistic synthetic data for car price prediction
np.random.seed(42)
n_cars = 1000

data_cars = {
    'Age': np.random.randint(1, 15, size=n_cars),
    'Mileage': np.random.randint(5000, 180000, size=n_cars),
    'Engine_Size': np.random.choice([1.0, 1.2, 1.4, 1.6, 2.0, 2.5, 3.0], size=n_cars),
    'Fuel_Type': np.random.choice(['Petrol', 'Diesel', 'Hybrid', 'Electric'], size=n_cars),
    'Transmission': np.random.choice(['Manual', 'Automatic'], size=n_cars)
}

df_cars = pd.DataFrame(data_cars)

# Programmatically assign base pricing derived from attributes plus noise
base_price = 30000
price = (
    base_price
    - (df_cars['Age'] * 1200)
    - (df_cars['Mileage'] * 0.08)
    + (df_cars['Engine_Size'] * 3500)
    + (df_cars['Fuel_Type'].map({'Petrol': 0, 'Diesel': 1500, 'Hybrid': 4000, 'Electric': 8000}))
    + np.random.normal(0, 1500, size=n_cars) # random fluctuations
)

# Ensure price is at least a realistic minimum base value
df_cars['Price'] = np.clip(price, 1500, None)

# Preview generated dataset
display(df_cars.head())
print(f"\nSummary statistics of target price:\n{df_cars['Price'].describe()}")
```
