```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 1. Load the Iris dataset
iris = load_iris()
X_data = iris.data
y_data = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# 2. Split dataset into training and testing sets (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2, random_state=42)

# 3. Initialize and train the Gaussian Naïve Bayes Classifier
nb_classifier = GaussianNB()
nb_classifier.fit(X_train, y_train)

# 4. Make predictions on test data
y_pred = nb_classifier.predict(X_test)

# 5. Evaluate Performance
accuracy = accuracy_score(y_test, y_pred) * 100
cm = confusion_matrix(y_test, y_pred)

print(f"--- Naïve Bayes Classification Results ---")
print(f"Overall Accuracy: {accuracy:.2f}%\n")
print("Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=target_names))

# 6. Plot Confusion Matrix Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix - Naïve Bayes Classifier')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.grid(False)
plt.show()
```
